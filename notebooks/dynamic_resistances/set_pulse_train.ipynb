```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import scienceplots
import seaborn as sns

plt.style.use(["science","bright","grid"])
```

```python
df = pd.read_csv("../../data/raw/dynamic_resistances/set_pulse_train/set_pulse_train.csv")
r = pd.read_parquet("../../data/raw/dynamic_resistances/set_pulse_train/resistances_set_pulse_train.parquet").mean(axis=1).values
```

```python
t = df['t'].values*1e9
x = df['E0'].values*1e3
y = df['Observable'].values*1e-6
e = df['Error'].values*1e-6
```

```python
fig = plt.figure(dpi=300, layout='constrained')
fig.set_figwidth(fig.get_figwidth()*2)
fig.set_figheight(fig.get_figheight()*1)

ax  = fig.add_subplot(2,1,1)
ax2 = ax.twinx()
_   = ax.errorbar(t, y, e)
_   = ax2.plot(t, x, color='black', ls='--')
_   = ax.set_ylim(0)
_   = ax.set_xlim(0)
_   = ax.set_ylabel('$I$~[pA]')
_   = ax.set_xlabel('$t$~[ns]')
_   = ax2.set_ylim(0)
_   = ax2.grid(False)
_   = ax2.set_yticklabels("")

ax  = fig.add_subplot(2,1,2)
ax2 = ax.twinx()
_   = ax.plot(t, r)
_   = ax2.plot(t, x, color='black', ls='--')
_   = ax.set_ylim(4,26)
_   = ax.set_xlim(0)
_   = ax.set_ylabel('$R$~[M$\\Omega$]')
_   = ax.set_xlabel('$t$~[ns]')
_   = ax2.set_ylim(0)
_   = ax2.grid(False)
_   = ax2.set_yticklabels("")

fig.savefig("plots/set_pulse_train/pulse_train.svg", transparent=True, bbox_inches='tight')
```
